```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('amazon.xlsx')
```

```python
df
```

### Data Inspection

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df = df.reset_index()
```

```python
df
```

```python
df[['Country','City','State']] = df['Geography'].str.split(",", expand=True)
```

```python
df
```

#### 1. What are the total Sales, Quantity, and Profit?

```python
print("The total sales is", df['Sales'].sum().round(1))
print("The total Quanity is", df['Quantity'].sum().round(1))
print("The total Profit is", df['Profit'].sum().round(1))
```

#### 2. Which are the Top 10 highest selling products?

```python
q2 = df.groupby('Product Name').agg(Total_QTY=('Quantity','sum')).sort_values(by='Total_QTY',
                                                                             ascending=False).reset_index().head(10)
q2
```

```python
plt.figure(figsize=(7,3))
sns.barplot(data = q2, x = 'Product Name', y = 'Total_QTY', palette='viridis')
plt.xticks(rotation=90)
plt.show()
```

```python
q3 = df.groupby('Category').agg(Total_profit=('Profit','sum')).sort_values(by=
                                                                           'Total_profit',ascending=False).head(1).reset_index()
```

```python
q3
```

```python
print("The highest profit category is ", q3.iloc[0,0])
```
